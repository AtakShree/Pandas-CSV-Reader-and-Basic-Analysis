```python
import pandas as pd

print("\n========== TASK 1 — Read CSV & Inspect ==========")
df = pd.read_csv("C:/Users/Dell/Downloads/employees.csv", parse_dates=["join_date"])

print("\n========== Head (first 5 rows) ==========")
print(df.head())
 
print("\n========== Tail (last 5 rows) ==========")
print(df.tail())
 
print("\n========== Column Data Types ==========")
print(df.dtypes)
```

```python
print("========== TASK 2 — Summary Statistics ==========")
 
numeric_cols = ["age", "salary", "experience_years", "performance_score"]
 
print("\n========== Summary Statistics ==========")
print(df[numeric_cols].describe().round(2))
 
print("\n========== Mean ==========")
print(df[numeric_cols].mean().round(2))
 
print("\n========== Median ==========")
print(df[numeric_cols].median())
 
print("\n========== Min ==========")
print(df[numeric_cols].min())
 
print("\n========== Max ==========")
print(df[numeric_cols].max())
 
print("\n========== Count ==========")
print(df[numeric_cols].count())
 
print("\n========== Average Salary by Department ==========")
print(df.groupby("department")["salary"].mean().round(2).sort_values(ascending=False))
 
print("\n========== Average Performance Score by Gender ==========")
print(df.groupby("gender")["performance_score"].mean().round(2))
```

```python
print("\n========== TASK 3 — Filter, Select & Slice ==========")
 
# Filter: salary > 80,000
high_earners = df[df["salary"] > 80_000]
print(f"\n========== Employees with Salary > ₹80,000 ({len(high_earners)} records) ==========")
print(high_earners[["name", "department", "salary"]].to_string(index=False))
 
# Filter: Engineering department
engineers = df[df["department"] == "Engineering"]
print(f"\n========== Engineering Department ({len(engineers)} records) ==========")
print(engineers[["name", "age", "salary", "performance_score"]].to_string(index=False))
 
# Filter: High performers with good pay (multi-condition)
stars = df[(df["performance_score"] >= 90) & (df["salary"] >= 80_000)]
print(f"\n========== Stars: Performance ≥ 90 AND Salary ≥ ₹80,000 ({len(stars)} records) ==========")
print(stars[["name", "department", "salary", "performance_score"]].to_string(index=False))
 
# Select specific columns
print(f"\n========== Selected Columns: name, department, city ({len(df[['name', 'department', 'city']].head(8))} records) ==========")
print(df[["name", "department", "city"]].head(8).to_string(index=False))
 
# Slice rows 5–10 using iloc
print("\n========== Slice: rows 5–10 (iloc) ==========")
print(df.iloc[5:11].to_string(index=False))
 
# Slice: top 5 by salary
print("\n========== Top 5 Highest Paid Employees ==========")
print(df.nlargest(5, "salary")[["name", "department", "salary"]].to_string(index=False))
```

```python
print("\n========== TASK 4 — Save Filtered Results ==========")
 
# Save high earners to CSV
high_earners.to_csv("high_earners.csv", index=False)
print("\n========== Saved: high_earners.csv ==========")
 
# Save engineering dept to CSV
engineers.to_csv("engineering_dept.csv", index=False)
print("\n========== Saved: engineering_dept.csv ==========")
 
# Save star performers to CSV
stars.to_csv("star_performers.csv", index=False)
print("\n========== Saved: star_performers.csv ==========")
 
# Save summary stats to CSV
summary_stats = df.groupby("department").agg(
    headcount=("employee_id", "count"),
    avg_salary=("salary", "mean"),
    avg_performance=("performance_score", "mean"),
    avg_experience=("experience_years", "mean"),
).round(2).reset_index()
summary_stats.to_csv("department_summary.csv", index=False)
print("\n========== Saved: department_summary.csv ==========")
 
print("\n========== Department Summary Preview ==========")
print(summary_stats.to_string(index=False))
```
